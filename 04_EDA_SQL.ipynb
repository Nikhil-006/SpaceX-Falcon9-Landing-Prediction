```python
import sqlite3
import pandas as pd

print("SQLite imported successfully!")
```

```python
df = pd.read_csv("../data/spacex_cleaned.csv")

conn = sqlite3.connect("../data/spacex_database.db")

df.to_sql("SPACEX", conn, if_exists="replace", index=False)

print("Dataset successfully loaded into SQL database!")
```

```python
query = "SELECT * FROM SPACEX LIMIT 5"

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT COUNT(*) AS Total_Launches
FROM SPACEX;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Class,
    COUNT(*) AS Total_Launches
FROM SPACEX
GROUP BY Class;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    LaunchSite,
    COUNT(*) AS Total_Launches
FROM SPACEX
GROUP BY LaunchSite
ORDER BY Total_Launches DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    LaunchSite,
    COUNT(*) AS Total_Launches,
    SUM(Class) AS Successful_Landings,
    ROUND(AVG(Class) * 100, 2) AS Success_Rate
FROM SPACEX
GROUP BY LaunchSite
ORDER BY Success_Rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Orbit,
    COUNT(*) AS Total_Launches,
    SUM(Class) AS Successful_Landings,
    ROUND(AVG(Class) * 100, 2) AS Success_Rate
FROM SPACEX
GROUP BY Orbit
ORDER BY Success_Rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Year,
    COUNT(*) AS Total_Launches,
    SUM(Class) AS Successful_Landings,
    ROUND(AVG(Class) * 100, 2) AS Success_Rate
FROM SPACEX
GROUP BY Year
ORDER BY Year;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Class,
    COUNT(*) AS Total_Launches,
    ROUND(AVG(PayloadMass), 2) AS Average_Payload_Mass,
    ROUND(MIN(PayloadMass), 2) AS Minimum_Payload_Mass,
    ROUND(MAX(PayloadMass), 2) AS Maximum_Payload_Mass
FROM SPACEX
GROUP BY Class
ORDER BY Class;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    COUNT(*) AS Total_Launches,
    SUM(Class) AS Successful_Landings,
    COUNT(*) - SUM(Class) AS Unsuccessful_Landings,
    ROUND(AVG(Class) * 100, 2) AS Overall_Success_Rate
FROM SPACEX;
"""

pd.read_sql_query(query, conn)
```
