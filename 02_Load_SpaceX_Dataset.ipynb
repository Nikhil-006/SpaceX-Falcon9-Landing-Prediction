```python
import requests

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"

response = requests.get(url, timeout=60)

print("Status Code:", response.status_code)

if response.status_code == 200:
    with open("../data/spacex_dataset.csv", "wb") as f:
        f.write(response.content)

    print("Dataset downloaded successfully!")
else:
    print("Download failed:", response.status_code)
```

```python
import pandas as pd

# Load the SpaceX dataset
df = pd.read_csv("../data/spacex_dataset.csv")

# Display dataset information
print("Dataset Shape:", df.shape)

# Display first 5 rows
df.head()
```

```python
df.info()
```

```python
# Check missing values in every column
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values)
```

```python
# Check duplicate rows
print("Total Duplicate Rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
# Convert Date column into datetime format
df["Date"] = pd.to_datetime(df["Date"])

# Check data type
print(df["Date"].dtype)

# Display first five dates
df[["FlightNumber", "Date"]].head()
```

```python
# Extract year from launch date
df["Year"] = df["Date"].dt.year

# Display launch year
df[["FlightNumber", "Date", "Year"]].head()
```

```python
# Check unique landing outcomes
print("Landing Outcomes:")
print(df["Outcome"].value_counts())

print("\nLanding Pad Values:")
print(df["LandingPad"].value_counts(dropna=False))
```

```python
print("Landing Success Distribution:")
print(df["Class"].value_counts())

print("\nLanding Success Percentage:")
print(df["Class"].value_counts(normalize=True) * 100)
```

```python
# Save the cleaned dataset
df.to_csv("../data/spacex_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Final Dataset Shape:", df.shape)
```
